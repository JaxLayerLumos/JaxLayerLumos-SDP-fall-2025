# Triple Junction Solar Cells

Triple junction solar cells are a type of multi-junction PV device that can exceed the Shockley-Queisser (detailed balance) limit.
It is designed to maximize efficiency by capturing and converting a broader spectrum of sunlight into electricity.

The three layers consist of the following:
1. Top Layer (GaInP - Gallium Indium Phosphide):
GaInP has a high bandgap (~1.85 eV) and absorbs high-energy photons (primarily in the blue and green regions of the spectrum).
2. Middle Layer (GaAs - Gallium Arsenide):
GaAs has a medium bandgap (~1.42 eV) and absorbs photons in the red and near-infrared region.
3. Bottom Layer (Ge - Germanium):
Ge has a low bandgap (~0.67 eV) and absorbs photons in the infrared region.

First of all, we import required packages.

```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches

fig, ax = plt.subplots(figsize=(12, 5))

# --- PARAMETERS ---
N = 3  # number of layers to draw (example)
layer_width = 1.8
layer_height = 5
spacing = 0.3
start_x = 1

# --- PERFECT CONDUCTOR ---
ax.add_patch(patches.Rectangle((0,0), 0.8, layer_height, 
                               facecolor='black'))
ax.text(0.4, layer_height/2, 
        "PERFECTLY\nCONDUCTING\nBACKING", 
        color="white", fontsize=12, ha='center', va='center', rotation=90)

# --- LAYERS ---
for i in range(N):
    x = start_x + i*(layer_width + spacing)
    ax.add_patch(patches.Rectangle((x,0), layer_width, layer_height, 
                                   fill=False, linewidth=1.5))
    
    ax.text(x + layer_width/2, 3.5, f"t$_{i+1}$", fontsize=16, ha='center')
    ax.text(x + layer_width/2, 2.5, f"Layer {i+1}\nMaterial n$_{i+1}$",
            ha='center', fontsize=14)
    ax.text(x + layer_width/2, 1.2, r"($\varepsilon_{%d},\mu_{%d}$)" % (i+1, i+1),
            ha='center', fontsize=14)

# Ellipsis (for many layers)
ax.text(start_x + N*(layer_width+spacing) + 0.3, layer_height/2, "⋯",
        fontsize=40, ha='left', va='center')

# --- INCOMING/OUTGOING EM WAVES ---
# Coordinates for the interaction point
px = start_x + N*(layer_width + spacing) + 1.1
py = layer_height/2

# Incoming ray
ax.plot([px+2, px], [py+2, py], linewidth=2)
ax.text(px+2.1, py+2, "EM$_{in}$", fontsize=14)

# Outgoing ray
ax.plot([px, px+2], [py, py-2], linewidth=2)
ax.text(px+2.1, py-2, "EM$_{out}$", fontsize=14)

# Incident and outgoing angle arcs
arc1 = patches.Arc((px, py), 2.0, 2.0, theta1=180, theta2=225, linewidth=2)
arc2 = patches.Arc((px, py), 2.0, 2.0, theta1=-45, theta2=0, linewidth=2)
ax.add_patch(arc1)
ax.add_patch(arc2)

ax.text(px-0.4, py+0.9, r"$\theta_{in}$", fontsize=16)
ax.text(px-0.4, py-1.0, r"$\theta_{out}$", fontsize=16)

# Dashed horizontal line
ax.plot([px-1, px+1], [py, py], '--', color='black')

# --- Cleanup ---
ax.set_xlim(-0.5, px+3)
ax.set_ylim(-0.5, layer_height+0.5)
ax.axis('off')

plt.show()
```
